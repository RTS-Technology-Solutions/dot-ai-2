## 1. System Architecture

### 1.1 Dual-Layer Learning System

The DOT-AI 2.0 system implements **two distinct learning layers**:

#### Layer 1: Individual Learning (Within Generation)
- **Brain & Memory:** Each agent has a 50-memory sliding window tracking interactions
- **Reward System:** Actions generate positive/negative rewards (eating: +energy/10, combat: +damage/10)
- **DNA Growth:** Energy cascade system (energy → health → DNA at 10% conversion)
- **Decision Making:** Utility-based action selection using perception and memory

#### Layer 2: Model Learning (Cross-Generational)
- **Champion Tracking:** Top 3 performers per generation saved (fitness = lifetime × (1+offspring) × (1+reward/100))
- **Hall of Fame:** All-time top 10 DNA configurations preserved
- **DNA Generation Strategy:**
  - 60% - Clone champion DNA + small mutations (1-3 genes, ±1-5 points)
  - 30% - Weighted random based on champion gene patterns
  - 10% - Pure random exploration

### 1.2 DNA System

Each agent has a genetic profile with **100 starting DNA points** (expandable through eating):

**Sensory Genes:** vision_distance, food_detection, dot_detection, dna_strength_detection, nearby_dot_density  
**Action Genes:** movement_speed, attack, defend, seek_mate, replicate, eat  
**Brain Genes:** memory_capacity, learning_rate, decision_speed

**Key Innovation:** DNA points can grow beyond 100 through successful eating, creating compound genetic improvement.
